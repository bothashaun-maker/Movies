```python
# Single-cell startup. No image generation starts automatically.
from pathlib import Path
import sys, os, subprocess, json, time, urllib.request, hashlib, shutil, re, signal, fcntl
# Turn off SDXL to run only the upscalers without its 7.1 GB checkpoint.
LOAD_SDXL = True
ENABLE_SEEDVR2 = True
ENABLE_FACESWAP = True
AUTO_PREPARE_FACE_PACK = True
AUTO_SAVE_PREPARED_FACE_PACK = True
PACK_BUILDER_SOURCE = '# CPU-only model pack builder. Run once and save the outputs on Kaggle.\nAUTO_CREATE_DATASET = True  # Uses your existing Kaggle authentication to create a private Dataset.\nDATASET_OWNER = \'\'  # Usually detected from authenticated Kaggle API credentials.\nPACK = \'faceswap\'  # faceswap ~843 MB; standard ~11.1 GB; flux3 ~17.75 GB\nfrom pathlib import Path\nimport os, json, hashlib, shutil, time, urllib.request, urllib.error, fcntl\nfrom pathlib import Path\nimport hashlib\nMODEL=\'Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors\'\nMODEL_SHA=\'c9e3e68f89b8e38689e1097d4be4573cf308de4e3fd044c64ca697bdb4aa8bca\'\nMODEL_SIZE=7105348188\ndef model_assets(load_sdxl=True, enable_seedvr2=True, enable_flux3=False, enable_faceswap=False):\n assets=[(\'upscale_models\',\'RealESRGAN_x4plus.pth\',\'https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth\',\'4fa0d38905f75ac06eb49a7951b426670021be3018265fd191d2125df9d682f1\',67040989)]\n if load_sdxl:assets.append((\'checkpoints\',MODEL,\'https://huggingface.co/RunDiffusion/Juggernaut-XL-v9/resolve/main/\'+MODEL,MODEL_SHA,MODEL_SIZE))\n if enable_seedvr2:\n  prefix=\'https://huggingface.co/Comfy-Org/SeedVR2/resolve/df48879708206a403d2a61acd55578c2e80fd233/\'\n  assets.extend([\n   (\'diffusion_models\',\'seedvr2_3b_int8_convrot.safetensors\',prefix+\'diffusion_models/seedvr2_3b_int8_convrot.safetensors\',\'c3dec8bcc5916843a8a858572970597462e1f2dc598d6dfd818f6cd40f53a157\',3458259704),\n   (\'vae\',\'seedvr2_ema_vae_fp16.safetensors\',prefix+\'vae/seedvr2_ema_vae_fp16.safetensors\',\'20678548f420d98d26f11442d3528f8b8c94e57ee046ef93dbb7633da8612ca1\',501324814)])\n if enable_flux3:\n  prefix=\'https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/9d36114cf7a2e03a45c28f5cc4987d20ebfdb6c1/\'\n  assets.extend([\n   (\'diffusion_models\',\'flux3-image-klein-int8-convrot.safetensors\',prefix+\'diffusion_models/flux3-image-klein-int8-convrot.safetensors\',\'847d8933548afd93df4af2cac2498e70891bf28d9dae06d695f536992e168fe2\',5827874779),\n   (\'text_encoders\',\'qwen3vl_4b_int8_convrot.safetensors\',prefix+\'text_encoders/qwen3vl_4b_int8_convrot.safetensors\',\'32de113cbf8282b02cbff0794b637b9d6676da09314a1db5a30bcea9bf4ee01c\',4858559240),\n   (\'vae\',\'flux2-vae.safetensors\',prefix+\'vae/flux2-vae.safetensors\',\'868fe7b343cc8f3a19dbcfcafbc3d5f888802be3f89bd81b65b3621a066ce8f3\',336211292),\n   (\'text_encoders\',\'qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors\',prefix+\'text_encoders/qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors\',\'e5d85a7421d197bf5a12474511936ecabce19a833252f16e9310a983194a4458\',6724090780)])\n if enable_faceswap:\n  prefix=\'https://huggingface.co/datasets/Gourieff/ReActor/resolve/079e0e459cd4261bbcba8c71d47df54a4ded7642/models/\'\n  assets.extend([\n   (\'insightface\',\'inswapper_128.onnx\',prefix+\'inswapper_128.onnx\',\'e4a3f08c753cb72d04e10aa0f7dbe3deebbf39567d4ead6dce08e98aa49e16af\',554253681),\n   (\'insightface\',\'buffalo_l.zip\',prefix+\'buffalo_l.zip\',\'80ffe37d8a5940d59a7384c201a2a38d4741f2f3c51eef46ebb28218a7b0ca2f\',288621354)])\n return assets\n\n\ndef file_sha(path):\n h=hashlib.sha256()\n with open(path,\'rb\') as f:\n  for chunk in iter(lambda:f.read(8*1024*1024),b\'\'):h.update(chunk)\n return h.hexdigest()\n\ndef selected_assets(pack):\n if pack==\'faceswap\':return [a for a in model_assets(False,False,False,True) if a[0]==\'insightface\']\n if pack==\'standard\':return model_assets(True,True,False)\n if pack==\'flux3\':return [a for a in model_assets(False,False,True) if \'black-forest-labs\' in a[2]]\n raise ValueError("PACK must be \'standard\', \'flux3\', or \'faceswap\'")\n\ndef transfer(url,target,digest,size):\n target=Path(target);target.parent.mkdir(parents=True,exist_ok=True)\n if target.exists():\n  if target.stat().st_size==size and file_sha(target)==digest:\n   print(\'Verified; reused:\',target.name,flush=True);return\n  raise RuntimeError(\'Existing file failed verification; preserve or move it before retrying: \'+str(target))\n partial=target.with_name(target.name+\'.partial\')\n for attempt in range(3):\n  offset=partial.stat().st_size if partial.exists() else 0\n  if offset>=size:\n   if offset==size and file_sha(partial)==digest:partial.replace(target);return\n   partial.unlink();offset=0\n  if shutil.disk_usage(target.parent).free<size-offset+512*1024**2:raise RuntimeError(\'Insufficient free disk for \'+target.name)\n  req=urllib.request.Request(url,headers={\'Range\':f\'bytes={offset}-\'} if offset else {})\n  try:\n   with urllib.request.urlopen(req,timeout=120) as response:\n    resume=offset>0 and response.status==206\n    if resume and not response.headers.get(\'Content-Range\',\'\').startswith(f\'bytes {offset}-\'):raise RuntimeError(\'Unexpected range response\')\n    written=offset if resume else 0;reported=written\n    with open(partial,\'ab\' if resume else \'wb\') as f:\n     while True:\n      data=response.read(8*1024*1024)\n      if not data:break\n      f.write(data);written+=len(data)\n      if written-reported>=128*1024**2:\n       print(f\'{target.name}: {written/size:.0%}\',flush=True);reported=written\n   if partial.stat().st_size!=size or file_sha(partial)!=digest:raise RuntimeError(\'Incomplete download or checksum mismatch\')\n   partial.replace(target);return\n  except (OSError,urllib.error.URLError,RuntimeError) as e:\n   if attempt==2:raise RuntimeError(\'Download failed for \'+target.name+\'; rerun to retry its partial file.\') from e\n   print(\'Retrying\',target.name,type(e).__name__,flush=True);time.sleep(2)\n\ndef build_pack(pack,output_root=Path(\'/kaggle/working/image-studio-model-packs\'),input_root=Path(\'/kaggle/input\')):\n assets=selected_assets(pack)\n root=Path(output_root)/pack;root.mkdir(parents=True,exist_ok=True)\n lock=open(root/\'.builder.lock\',\'w\')\n try:fcntl.flock(lock,fcntl.LOCK_EX|fcntl.LOCK_NB)\n except BlockingIOError:lock.close();raise RuntimeError(\'This pack is already being built.\')\n try:\n  manifest=root/\'manifest.json\'\n  # An interrupted rebuild must never leave a completed manifest behind.\n  if manifest.exists():manifest.replace(root/\'manifest.previous.json\')\n  print(\'Building\',pack,\'on CPU; model bytes:\',sum(a[4] for a in assets),flush=True)\n  records=[]\n  for folder,name,url,digest,size in assets:\n   target=root/name;target.parent.mkdir(parents=True,exist_ok=True)\n   # Saved packs must contain real files: input symlinks would not be a durable output.\n   if not target.exists() and Path(input_root).exists():\n    for candidate in Path(input_root).rglob(name):\n     if candidate.is_file() and candidate.stat().st_size==size and file_sha(candidate)==digest:\n      if shutil.disk_usage(target.parent).free<size+512*1024**2:raise RuntimeError(\'Insufficient disk to copy saved model\')\n      tmp=target.with_name(name+\'.partial\');shutil.copyfile(candidate,tmp);tmp.replace(target);print(\'Copied saved input:\',name,flush=True);break\n   if not target.exists() and \'black-forest-labs\' in url:\n    if shutil.disk_usage(target.parent).free<size+512*1024**2:raise RuntimeError(\'Insufficient free disk for FLUX3 asset\')\n    try:\n     from kaggle_secrets import UserSecretsClient\n     token=UserSecretsClient().get_secret(\'HF_TOKEN\')\n    except Exception:raise RuntimeError(\'Accept FLUX3 terms on Hugging Face and enable Kaggle Secret HF_TOKEN.\') from None\n    try:\n     from huggingface_hub import hf_hub_download\n     saved=hf_hub_download(repo_id=\'black-forest-labs/flux-3-image-klein-comfy\',filename=folder+\'/\'+name,revision=\'9d36114cf7a2e03a45c28f5cc4987d20ebfdb6c1\',local_dir=root/\'models\',token=token)\n     Path(saved).replace(target)\n    finally:del token\n   transfer(url,target,digest,size)\n   if target.is_symlink() or target.stat().st_size!=size or file_sha(target)!=digest:raise RuntimeError(\'Pack verification failed: \'+name)\n   records.append({\'path\':target.relative_to(root).as_posix(),\'bytes\':size,\'sha256\':digest,\'source\':url})\n  data={\'schema\':1,\'pack\':pack,\'complete\':True,\'model_bytes\':sum(a[4] for a in assets),\'files\':records}\n  tmp=manifest.with_suffix(\'.json.partial\');tmp.write_text(json.dumps(data,indent=2));tmp.replace(manifest)\n  print(\'\\nPACK COMPLETE:\',root,flush=True)\n  print(\'Save Version with outputs. In Image Studio v4: Add Input → Notebook Output Files → select this saved version.\',flush=True)\n  print(\'No GPU or image generation was used. Keep model packs private and follow model licenses.\',flush=True)\n  return data\n finally:fcntl.flock(lock,fcntl.LOCK_UN);lock.close()\n\ndef save_private_dataset(pack_root,owner=\'\',api=None):\n root=Path(pack_root);manifest=json.loads((root/\'manifest.json\').read_text())\n if not manifest.get(\'complete\'):raise RuntimeError(\'Refusing to upload an incomplete pack.\')\n for record in manifest[\'files\']:\n  p=root/record[\'path\']\n  if p.is_symlink() or not p.is_file() or p.stat().st_size!=record[\'bytes\'] or file_sha(p)!=record[\'sha256\']:raise RuntimeError(\'Model verification failed before Dataset upload: \'+p.name)\n if api is None:\n  # Kaggle API uses private notebook credentials. Never print or store their values.\n  if not os.environ.get(\'KAGGLE_API_TOKEN\') and not os.environ.get(\'KAGGLE_KEY\'):\n   try:\n    from kaggle_secrets import UserSecretsClient\n    os.environ[\'KAGGLE_API_TOKEN\']=UserSecretsClient().get_secret(\'KAGGLE_API_TOKEN\')\n   except Exception:pass\n  from kaggle.api.kaggle_api_extended import KaggleApi\n  api=KaggleApi();api.authenticate()\n owner=owner or api.config_values.get(\'username\',\'\')\n import re\n if not re.fullmatch(r\'[A-Za-z0-9_-]+\',owner):raise RuntimeError(\'Set DATASET_OWNER to your Kaggle username or configure Kaggle authentication privately.\')\n fingerprint=hashlib.sha256(json.dumps(manifest[\'files\'],sort_keys=True).encode()).hexdigest()[:12]\n slug=\'image-studio-\'+manifest[\'pack\']+\'-\'+fingerprint\n handle=owner+\'/\'+slug;url=\'https://www.kaggle.com/datasets/\'+handle\n receipt=root/\'dataset-upload.json\'\n if receipt.exists() and json.loads(receipt.read_text()).get(\'handle\')==handle:\n  print(\'Previously submitted Dataset:\',url,flush=True);return url\n metadata={\'title\':\'Image Studio \'+manifest[\'pack\']+\' \'+fingerprint,\'id\':handle,\'licenses\':[{\'name\':\'other\'}],\'description\':\'Private verified model pack for Image Studio. Original model licenses apply; no relicensing. See manifest.json for model sources and SHA256 values.\'}\n (root/\'dataset-metadata.json\').write_text(json.dumps(metadata,indent=2))\n # Files are flat: never zip many GiB or omit nested weights by accident.\n # create_new(public=False) will not modify an existing dataset or its visibility.\n result=api.dataset_create_new(str(root),public=False,quiet=False,convert_to_csv=False,dir_mode=\'skip\')\n error=result.get(\'error\') if isinstance(result,dict) else getattr(result,\'error\',None)\n if error:raise RuntimeError(\'Dataset creation was not confirmed: \'+str(error)+\'. Model files remain saved; existing datasets were not changed.\')\n receipt.write_text(json.dumps({\'handle\':handle,\'url\':url,\'submitted\':True,\'requested_private\':True,\'manifest_fingerprint\':fingerprint},indent=2))\n print(\'PRIVATE DATASET SUBMITTED:\',url,flush=True)\n print(\'Wait for Kaggle processing, then Image Studio v4 → Add Input → Your Datasets. No PC download needed.\',flush=True)\n return url\n\nif __name__==\'__main__\':\n build_pack(PACK)\n if AUTO_CREATE_DATASET:\n  try:save_private_dataset(Path(\'/kaggle/working/image-studio-model-packs\')/PACK,DATASET_OWNER)\n  except (Exception,SystemExit):\n   print(\'Automatic Dataset upload did not complete. The verified pack is retained in notebook outputs. Check Kaggle API authentication privately (KAGGLE_API_TOKEN secret), set DATASET_OWNER if needed, then rerun. No credentials are printed.\',flush=True)\n   raise\n\n'
"""Resolve saved image models and repair only the optional face-pack dependency."""
from pathlib import Path
import shutil

def model_candidates(folder,name,input_root=Path('/kaggle/input'),runtime_root=Path('/kaggle/temp'),pack_root=Path('/kaggle/working/image-studio-model-packs')):
 paths=[]
 for root in [Path(input_root),Path(pack_root)]:
  if root.exists():paths.extend(root.rglob(name))
 paths.extend(Path(runtime_root)/('image-studio-v'+str(v))/'ComfyUI/models'/folder/name for v in range(1,6))
 return [p for p in paths if p.is_file()]

def prepare_startup_models(assets,builder_source,require_attached=True,prepare_faces=True,save_face_dataset=True,input_root=Path('/kaggle/input'),runtime_root=Path('/kaggle/temp'),pack_root=Path('/kaggle/working/image-studio-model-packs'),builder=None):
 """Do not initialize a GPU or fetch existing generation-model families."""
 input_root=Path(input_root);runtime_root=Path(runtime_root);pack_root=Path(pack_root)
 def present(a):return [p for p in model_candidates(a[0],a[1],input_root,runtime_root,pack_root) if p.stat().st_size==a[4]]
 missing=[a for a in assets if not present(a)]
 missing_core=[a for a in missing if a[0]!='insightface']
 missing_face=[a for a in missing if a[0]=='insightface']
 if require_attached and missing_core:
  raise RuntimeError('Attach your saved standard/FLUX3 model pack. Missing: '+', '.join(a[1] for a in missing_core)+'. No generation-model downloads started.')
 if require_attached and missing_face and not prepare_faces:
  raise RuntimeError('Attach the faceswap model pack or enable AUTO_PREPARE_FACE_PACK. Missing: '+', '.join(a[1] for a in missing_face))
 owns_manifest=(pack_root/'faceswap/manifest.json').is_file()
 if prepare_faces and (missing_face or (owns_manifest and any(a[0]=='insightface' for a in assets))):
  if builder is None:
   builder={'__name__':'image_studio_pack_support'}
   exec(compile(builder_source,'embedded_image_pack_builder','exec'),builder)
  if missing_face:
   print('Preparing missing face-swap models on CPU (up to 843 MB). Existing generation packs are reused.',flush=True)
   output=pack_root/'faceswap';output.mkdir(parents=True,exist_ok=True)
   for a in assets:
    if a[0]!='insightface':continue
    dest=output/a[1]
    if dest.exists():continue
    for candidate in present(a):
     if builder['file_sha'](candidate)==a[3]:
      if shutil.disk_usage(output).free<a[4]+512*1024**2:raise RuntimeError('Insufficient free disk for the saved face pack.')
      temporary=dest.with_name(dest.name+'.partial');shutil.copyfile(candidate,temporary);temporary.replace(dest);break
   builder['build_pack']('faceswap',output_root=pack_root,input_root=input_root)
  if save_face_dataset:
   try:builder['save_private_dataset'](pack_root/'faceswap')
   except (Exception,SystemExit) as error:
    print('Face models are ready for this session, but private Dataset submission did not complete ('+type(error).__name__+').',flush=True)
    print('Verified files remain at '+str(pack_root/'faceswap')+'. Save Version with outputs or retry the model-pack builder to keep them for future sessions.',flush=True)
  else:print('Face pack prepared. Save Version with outputs to preserve it.',flush=True)
 # Size discovery is only preflight. The renderer still verifies model SHA256 before use.
 return {a[1]:present(a) for a in assets}

FACE_NODE_SOURCE = '"""Visible InsightFace detection and swapping stages for Image Studio."""\nfrom pathlib import Path\nfrom functools import lru_cache\nimport os\nimport numpy as np\nimport torch\nimport folder_paths\n\nROOT=Path(folder_paths.models_dir)/\'insightface\'\n\ndef image_bgr(image):\n if image.ndim!=4 or image.shape[0]!=1 or image.shape[-1]!=3:\n  raise ValueError(\'Use one RGB source photo and one RGB target photo, not image batches.\')\n return np.ascontiguousarray((image[0].detach().cpu().numpy().clip(0,1)*255).round().astype(np.uint8)[:,:,::-1])\n\n@lru_cache(maxsize=1)\ndef analyser():\n from insightface.app import FaceAnalysis\n import onnxruntime as ort\n models=ROOT/\'models/buffalo_l\'\n for name in [\'det_10g.onnx\',\'w600k_r50.onnx\']:\n  if not (models/name).is_file():raise RuntimeError(\'Attach the face-swap model pack and rerun studio startup; missing \'+name)\n options=ort.SessionOptions();options.intra_op_num_threads=min(4,os.cpu_count() or 1)\n app=FaceAnalysis(name=\'buffalo_l\',root=str(ROOT),allowed_modules=[\'detection\',\'recognition\'],providers=[\'CPUExecutionProvider\'],sess_options=options)\n app.prepare(ctx_id=-1,det_size=(640,640))\n return app\n\n@lru_cache(maxsize=1)\ndef swapper():\n from insightface.model_zoo import get_model\n import onnxruntime as ort\n path=ROOT/\'inswapper_128.onnx\'\n if not path.is_file():raise RuntimeError(\'Missing inswapper_128.onnx; attach the face-swap model pack.\')\n options=ort.SessionOptions();options.intra_op_num_threads=min(4,os.cpu_count() or 1)\n return get_model(str(path),providers=[\'CPUExecutionProvider\'],sess_options=options)\n\nclass StudioFaceDetect:\n @classmethod\n def INPUT_TYPES(cls):return {\'required\':{\'image\':(\'IMAGE\',)}}\n RETURN_TYPES=(\'STUDIO_FACES\',\'STRING\')\n RETURN_NAMES=(\'faces\',\'detected_faces\')\n FUNCTION=\'detect\';CATEGORY=\'Image Studio/Face Swap\'\n def detect(self,image):\n  bgr=image_bgr(image)\n  faces=sorted(analyser().get(bgr),key=lambda f:float((f.bbox[0]+f.bbox[2])/2))\n  if not faces:raise ValueError(\'No face detected. Use a clear, larger, unobstructed face photo.\')\n  descriptions=[f\'{i}: box \'+\',\'.join(str(int(v)) for v in f.bbox) for i,f in enumerate(faces)]\n  data={\'faces\':faces,\'image\':bgr,\'shape\':bgr.shape}\n  return data,\'Faces numbered left to right, starting at 0. \'+ \'; \'.join(descriptions)\n\nclass StudioFaceSwap:\n @classmethod\n def INPUT_TYPES(cls):\n  return {\'required\':{\'source_faces\':(\'STUDIO_FACES\',),\'target_faces\':(\'STUDIO_FACES\',),\'source_index\':(\'INT\',{\'default\':0,\'min\':0,\'max\':99}),\'target_index\':(\'INT\',{\'default\':0,\'min\':0,\'max\':99})}}\n RETURN_TYPES=(\'IMAGE\',);FUNCTION=\'swap\';CATEGORY=\'Image Studio/Face Swap\'\n def swap(self,source_faces,target_faces,source_index,target_index):\n  if not 0<=source_index<len(source_faces[\'faces\']):raise ValueError(\'Source face index is outside the detected face list.\')\n  if not 0<=target_index<len(target_faces[\'faces\']):raise ValueError(\'Target face index is outside the detected face list.\')\n  result=swapper().get(target_faces[\'image\'].copy(),target_faces[\'faces\'][target_index],source_faces[\'faces\'][source_index],paste_back=True)\n  if result.shape!=target_faces[\'shape\'] or not np.isfinite(result).all():raise RuntimeError(\'Invalid face-swap output.\')\n  rgb=np.ascontiguousarray(result[:,:,::-1]);return (torch.from_numpy(rgb.astype(np.float32)/255.0).unsqueeze(0),)\n\nNODE_CLASS_MAPPINGS={\'StudioFaceDetect\':StudioFaceDetect,\'StudioFaceSwap\':StudioFaceSwap}\nNODE_DISPLAY_NAME_MAPPINGS={\'StudioFaceDetect\':\'Detect Faces — Image Studio\',\'StudioFaceSwap\':\'Swap Selected Face — Image Studio\'}\n'
# Optional gated model: accept its terms on Hugging Face and add Kaggle Secret HF_TOKEN.
ENABLE_FLUX3 = False
REQUIRE_ATTACHED_MODELS = True
# The builder embeds the shared asset catalog here.
from pathlib import Path
import hashlib
MODEL='Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors'
MODEL_SHA='c9e3e68f89b8e38689e1097d4be4573cf308de4e3fd044c64ca697bdb4aa8bca'
MODEL_SIZE=7105348188
def model_assets(load_sdxl=True, enable_seedvr2=True, enable_flux3=False, enable_faceswap=False):
 assets=[('upscale_models','RealESRGAN_x4plus.pth','https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth','4fa0d38905f75ac06eb49a7951b426670021be3018265fd191d2125df9d682f1',67040989)]
 if load_sdxl:assets.append(('checkpoints',MODEL,'https://huggingface.co/RunDiffusion/Juggernaut-XL-v9/resolve/main/'+MODEL,MODEL_SHA,MODEL_SIZE))
 if enable_seedvr2:
  prefix='https://huggingface.co/Comfy-Org/SeedVR2/resolve/df48879708206a403d2a61acd55578c2e80fd233/'
  assets.extend([
   ('diffusion_models','seedvr2_3b_int8_convrot.safetensors',prefix+'diffusion_models/seedvr2_3b_int8_convrot.safetensors','c3dec8bcc5916843a8a858572970597462e1f2dc598d6dfd818f6cd40f53a157',3458259704),
   ('vae','seedvr2_ema_vae_fp16.safetensors',prefix+'vae/seedvr2_ema_vae_fp16.safetensors','20678548f420d98d26f11442d3528f8b8c94e57ee046ef93dbb7633da8612ca1',501324814)])
 if enable_flux3:
  prefix='https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/9d36114cf7a2e03a45c28f5cc4987d20ebfdb6c1/'
  assets.extend([
   ('diffusion_models','flux3-image-klein-int8-convrot.safetensors',prefix+'diffusion_models/flux3-image-klein-int8-convrot.safetensors','847d8933548afd93df4af2cac2498e70891bf28d9dae06d695f536992e168fe2',5827874779),
   ('text_encoders','qwen3vl_4b_int8_convrot.safetensors',prefix+'text_encoders/qwen3vl_4b_int8_convrot.safetensors','32de113cbf8282b02cbff0794b637b9d6676da09314a1db5a30bcea9bf4ee01c',4858559240),
   ('vae','flux2-vae.safetensors',prefix+'vae/flux2-vae.safetensors','868fe7b343cc8f3a19dbcfcafbc3d5f888802be3f89bd81b65b3621a066ce8f3',336211292),
   ('text_encoders','qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors',prefix+'text_encoders/qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors','e5d85a7421d197bf5a12474511936ecabce19a833252f16e9310a983194a4458',6724090780)])
 if enable_faceswap:
  prefix='https://huggingface.co/datasets/Gourieff/ReActor/resolve/079e0e459cd4261bbcba8c71d47df54a4ded7642/models/'
  assets.extend([
   ('insightface','inswapper_128.onnx',prefix+'inswapper_128.onnx','e4a3f08c753cb72d04e10aa0f7dbe3deebbf39567d4ead6dce08e98aa49e16af',554253681),
   ('insightface','buffalo_l.zip',prefix+'buffalo_l.zip','80ffe37d8a5940d59a7384c201a2a38d4741f2f3c51eef46ebb28218a7b0ca2f',288621354)])
 return assets

WORKFLOW_PAYLOAD = {'Create': {'last_node_id': 7, 'last_link_id': 9, 'nodes': [{'id': 1, 'type': 'CheckpointLoaderSimple', 'pos': [0, 0], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [3]}, {'name': 'CLIP', 'type': 'CLIP', 'links': [1, 2]}, {'name': 'VAE', 'type': 'VAE', 'links': [8]}], 'properties': {'Node name for S&R': 'CheckpointLoaderSimple'}, 'widgets_values': ['Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors']}, {'id': 2, 'type': 'CLIPTextEncode', 'pos': [380, 0], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 1}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [4]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['Cinematic photograph of a lone explorer overlooking a vast alien coastline, dramatic clouds, natural light, detailed textures, thoughtful composition'], 'title': 'Describe your image'}, {'id': 3, 'type': 'CLIPTextEncode', 'pos': [380, 300], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 2}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [5]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['blurry, low quality, distorted anatomy, watermark, oversaturated'], 'title': 'Negative prompt'}, {'id': 4, 'type': 'EmptyLatentImage', 'pos': [380, 650], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [6]}], 'properties': {'Node name for S&R': 'EmptyLatentImage'}, 'widgets_values': [1024, 1024, 1]}, {'id': 5, 'type': 'KSampler', 'pos': [770, 0], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 3}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 4}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 5}, {'name': 'latent_image', 'type': 'LATENT', 'link': 6}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [7]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'randomize', 30, 5.5, 'dpmpp_2m', 'karras', 1.0]}, {'id': 6, 'type': 'VAEDecodeTiled', 'pos': [1130, 350], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 7}, {'name': 'vae', 'type': 'VAE', 'link': 8}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [9]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 64, 64, 8]}, {'id': 7, 'type': 'SaveImage', 'pos': [1490, 350], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 9}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Create']}], 'links': [[1, 1, 1, 2, 0, 'CLIP'], [2, 1, 1, 3, 0, 'CLIP'], [3, 1, 0, 5, 0, 'MODEL'], [4, 2, 0, 5, 1, 'CONDITIONING'], [5, 3, 0, 5, 2, 'CONDITIONING'], [6, 4, 0, 5, 3, 'LATENT'], [7, 5, 0, 6, 0, 'LATENT'], [8, 1, 2, 6, 1, 'VAE'], [9, 6, 0, 7, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Restyle': {'last_node_id': 8, 'last_link_id': 11, 'nodes': [{'id': 1, 'type': 'CheckpointLoaderSimple', 'pos': [0, 0], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [5]}, {'name': 'CLIP', 'type': 'CLIP', 'links': [1, 2]}, {'name': 'VAE', 'type': 'VAE', 'links': [4, 10]}], 'properties': {'Node name for S&R': 'CheckpointLoaderSimple'}, 'widgets_values': ['Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors']}, {'id': 2, 'type': 'CLIPTextEncode', 'pos': [380, 0], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 1}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [6]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['Cinematic photograph of a lone explorer overlooking a vast alien coastline, dramatic clouds, natural light, detailed textures, thoughtful composition'], 'title': 'Describe your image'}, {'id': 3, 'type': 'CLIPTextEncode', 'pos': [380, 300], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 2}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [7]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['blurry, low quality, distorted anatomy, watermark, oversaturated'], 'title': 'Negative prompt'}, {'id': 4, 'type': 'LoadImage', 'pos': [0, 400], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [3]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload a source image'}, {'id': 5, 'type': 'VAEEncode', 'pos': [380, 650], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'pixels', 'type': 'IMAGE', 'link': 3}, {'name': 'vae', 'type': 'VAE', 'link': 4}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [8]}], 'properties': {'Node name for S&R': 'VAEEncode'}, 'widgets_values': []}, {'id': 6, 'type': 'KSampler', 'pos': [770, 0], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 5}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 6}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 7}, {'name': 'latent_image', 'type': 'LATENT', 'link': 8}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [9]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'randomize', 30, 5.5, 'dpmpp_2m', 'karras', 0.6]}, {'id': 7, 'type': 'VAEDecodeTiled', 'pos': [1130, 350], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 9}, {'name': 'vae', 'type': 'VAE', 'link': 10}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [11]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 64, 64, 8]}, {'id': 8, 'type': 'SaveImage', 'pos': [1490, 350], 'size': [320, 250], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 11}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Restyle']}], 'links': [[1, 1, 1, 2, 0, 'CLIP'], [2, 1, 1, 3, 0, 'CLIP'], [3, 4, 0, 5, 0, 'IMAGE'], [4, 1, 2, 5, 1, 'VAE'], [5, 1, 0, 6, 0, 'MODEL'], [6, 2, 0, 6, 1, 'CONDITIONING'], [7, 3, 0, 6, 2, 'CONDITIONING'], [8, 5, 0, 6, 3, 'LATENT'], [9, 6, 0, 7, 0, 'LATENT'], [10, 1, 2, 7, 1, 'VAE'], [11, 7, 0, 8, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Inpaint': {'last_node_id': 8, 'last_link_id': 12, 'nodes': [{'id': 1, 'type': 'CheckpointLoaderSimple', 'pos': [0, 0], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [6]}, {'name': 'CLIP', 'type': 'CLIP', 'links': [1, 2]}, {'name': 'VAE', 'type': 'VAE', 'links': [5, 11]}], 'properties': {'Node name for S&R': 'CheckpointLoaderSimple'}, 'widgets_values': ['Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors']}, {'id': 2, 'type': 'CLIPTextEncode', 'pos': [380, 0], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 1}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [7]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['Cinematic photograph of a lone explorer overlooking a vast alien coastline, dramatic clouds, natural light, detailed textures, thoughtful composition'], 'title': 'Describe your image'}, {'id': 3, 'type': 'CLIPTextEncode', 'pos': [380, 300], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 2}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [8]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['blurry, low quality, distorted anatomy, watermark, oversaturated'], 'title': 'Negative prompt'}, {'id': 4, 'type': 'LoadImage', 'pos': [0, 400], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [4]}, {'name': 'MASK', 'type': 'MASK', 'links': [3]}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload image; right-click to open MaskEditor'}, {'id': 5, 'type': 'VAEEncodeForInpaint', 'pos': [380, 650], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'pixels', 'type': 'IMAGE', 'link': 4}, {'name': 'vae', 'type': 'VAE', 'link': 5}, {'name': 'mask', 'type': 'MASK', 'link': 3}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [9]}], 'properties': {'Node name for S&R': 'VAEEncodeForInpaint'}, 'widgets_values': [6]}, {'id': 6, 'type': 'KSampler', 'pos': [770, 0], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 6}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 7}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 8}, {'name': 'latent_image', 'type': 'LATENT', 'link': 9}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [10]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'randomize', 30, 5.5, 'dpmpp_2m', 'karras', 1.0]}, {'id': 7, 'type': 'VAEDecodeTiled', 'pos': [1130, 350], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 10}, {'name': 'vae', 'type': 'VAE', 'link': 11}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [12]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 64, 64, 8]}, {'id': 8, 'type': 'SaveImage', 'pos': [1490, 350], 'size': [320, 250], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 12}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Inpaint']}], 'links': [[1, 1, 1, 2, 0, 'CLIP'], [2, 1, 1, 3, 0, 'CLIP'], [3, 4, 1, 5, 2, 'MASK'], [4, 4, 0, 5, 0, 'IMAGE'], [5, 1, 2, 5, 1, 'VAE'], [6, 1, 0, 6, 0, 'MODEL'], [7, 2, 0, 6, 1, 'CONDITIONING'], [8, 3, 0, 6, 2, 'CONDITIONING'], [9, 5, 0, 6, 3, 'LATENT'], [10, 6, 0, 7, 0, 'LATENT'], [11, 1, 2, 7, 1, 'VAE'], [12, 7, 0, 8, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Detail': {'last_node_id': 9, 'last_link_id': 14, 'nodes': [{'id': 1, 'type': 'CheckpointLoaderSimple', 'pos': [0, 0], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [3, 8]}, {'name': 'CLIP', 'type': 'CLIP', 'links': [1, 2]}, {'name': 'VAE', 'type': 'VAE', 'links': [13]}], 'properties': {'Node name for S&R': 'CheckpointLoaderSimple'}, 'widgets_values': ['Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors']}, {'id': 2, 'type': 'CLIPTextEncode', 'pos': [380, 0], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 1}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [4, 9]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['Cinematic photograph of a lone explorer overlooking a vast alien coastline, dramatic clouds, natural light, detailed textures, thoughtful composition'], 'title': 'Describe your image'}, {'id': 3, 'type': 'CLIPTextEncode', 'pos': [380, 300], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 2}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [5, 10]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['blurry, low quality, distorted anatomy, watermark, oversaturated'], 'title': 'Negative prompt'}, {'id': 4, 'type': 'EmptyLatentImage', 'pos': [380, 650], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [6]}], 'properties': {'Node name for S&R': 'EmptyLatentImage'}, 'widgets_values': [1024, 1024, 1]}, {'id': 5, 'type': 'KSampler', 'pos': [770, 0], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 3}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 4}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 5}, {'name': 'latent_image', 'type': 'LATENT', 'link': 6}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [7]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'randomize', 30, 5.5, 'dpmpp_2m', 'karras', 1.0]}, {'id': 6, 'type': 'LatentUpscaleBy', 'pos': [1130, 0], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 7}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [11]}], 'properties': {'Node name for S&R': 'LatentUpscaleBy'}, 'widgets_values': ['bislerp', 1.5]}, {'id': 7, 'type': 'KSampler', 'pos': [1490, 0], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 8}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 9}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 10}, {'name': 'latent_image', 'type': 'LATENT', 'link': 11}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [12]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'fixed', 20, 5.0, 'dpmpp_2m', 'karras', 0.3], 'title': 'Detail pass — 1536px; costs more VRAM'}, {'id': 8, 'type': 'VAEDecodeTiled', 'pos': [1900, 350], 'size': [320, 250], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 12}, {'name': 'vae', 'type': 'VAE', 'link': 13}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [14]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 64, 64, 8]}, {'id': 9, 'type': 'SaveImage', 'pos': [2260, 350], 'size': [320, 250], 'flags': {}, 'order': 8, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 14}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Detail']}], 'links': [[1, 1, 1, 2, 0, 'CLIP'], [2, 1, 1, 3, 0, 'CLIP'], [3, 1, 0, 5, 0, 'MODEL'], [4, 2, 0, 5, 1, 'CONDITIONING'], [5, 3, 0, 5, 2, 'CONDITIONING'], [6, 4, 0, 5, 3, 'LATENT'], [7, 5, 0, 6, 0, 'LATENT'], [8, 1, 0, 7, 0, 'MODEL'], [9, 2, 0, 7, 1, 'CONDITIONING'], [10, 3, 0, 7, 2, 'CONDITIONING'], [11, 6, 0, 7, 3, 'LATENT'], [12, 7, 0, 8, 0, 'LATENT'], [13, 1, 2, 8, 1, 'VAE'], [14, 8, 0, 9, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Upscale 2x': {'last_node_id': 5, 'last_link_id': 4, 'nodes': [{'id': 1, 'type': 'LoadImage', 'pos': [0, 100], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [2]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload an image to enlarge'}, {'id': 2, 'type': 'UpscaleModelLoader', 'pos': [0, 440], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'UPSCALE_MODEL', 'type': 'UPSCALE_MODEL', 'links': [1]}], 'properties': {'Node name for S&R': 'UpscaleModelLoader'}, 'widgets_values': ['RealESRGAN_x4plus.pth']}, {'id': 3, 'type': 'ImageUpscaleWithModel', 'pos': [360, 100], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'upscale_model', 'type': 'UPSCALE_MODEL', 'link': 1}, {'name': 'image', 'type': 'IMAGE', 'link': 2}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [3]}], 'properties': {'Node name for S&R': 'ImageUpscaleWithModel'}, 'widgets_values': []}, {'id': 4, 'type': 'ImageScaleBy', 'pos': [720, 100], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [{'name': 'image', 'type': 'IMAGE', 'link': 3}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [4]}], 'properties': {'Node name for S&R': 'ImageScaleBy'}, 'widgets_values': ['lanczos', 0.5], 'title': '4x model then downsample to 2x'}, {'id': 5, 'type': 'SaveImage', 'pos': [1080, 100], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 4}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Upscale 2x']}], 'links': [[1, 2, 0, 3, 0, 'UPSCALE_MODEL'], [2, 1, 0, 3, 1, 'IMAGE'], [3, 3, 0, 4, 0, 'IMAGE'], [4, 4, 0, 5, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Upscale 4x': {'last_node_id': 4, 'last_link_id': 3, 'nodes': [{'id': 1, 'type': 'LoadImage', 'pos': [0, 100], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [2]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload an image to enlarge'}, {'id': 2, 'type': 'UpscaleModelLoader', 'pos': [0, 440], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'UPSCALE_MODEL', 'type': 'UPSCALE_MODEL', 'links': [1]}], 'properties': {'Node name for S&R': 'UpscaleModelLoader'}, 'widgets_values': ['RealESRGAN_x4plus.pth']}, {'id': 3, 'type': 'ImageUpscaleWithModel', 'pos': [360, 100], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'upscale_model', 'type': 'UPSCALE_MODEL', 'link': 1}, {'name': 'image', 'type': 'IMAGE', 'link': 2}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [3]}], 'properties': {'Node name for S&R': 'ImageUpscaleWithModel'}, 'widgets_values': []}, {'id': 4, 'type': 'SaveImage', 'pos': [1080, 100], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 3}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Upscale 4x']}], 'links': [[1, 2, 0, 3, 0, 'UPSCALE_MODEL'], [2, 1, 0, 3, 1, 'IMAGE'], [3, 3, 0, 4, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Upscale Creative': {'last_node_id': 7, 'last_link_id': 9, 'nodes': [{'id': 1, 'type': 'CheckpointLoaderSimple', 'pos': [0, 0], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [4]}, {'name': 'CLIP', 'type': 'CLIP', 'links': [1, 2]}, {'name': 'VAE', 'type': 'VAE', 'links': [7]}], 'properties': {'Node name for S&R': 'CheckpointLoaderSimple'}, 'widgets_values': ['Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors']}, {'id': 2, 'type': 'CLIPTextEncode', 'pos': [380, 0], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 1}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [5]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['Cinematic photograph of a lone explorer overlooking a vast alien coastline, dramatic clouds, natural light, detailed textures, thoughtful composition'], 'title': 'Describe your image'}, {'id': 3, 'type': 'CLIPTextEncode', 'pos': [380, 300], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'clip', 'type': 'CLIP', 'link': 2}], 'outputs': [{'name': 'CONDITIONING', 'type': 'CONDITIONING', 'links': [6]}], 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['blurry, low quality, distorted anatomy, watermark, oversaturated'], 'title': 'Negative prompt'}, {'id': 4, 'type': 'LoadImage', 'pos': [0, 400], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [3]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload source; 2x creative enhancement'}, {'id': 5, 'type': 'UpscaleModelLoader', 'pos': [380, 650], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'UPSCALE_MODEL', 'type': 'UPSCALE_MODEL', 'links': [8]}], 'properties': {'Node name for S&R': 'UpscaleModelLoader'}, 'widgets_values': ['RealESRGAN_x4plus.pth']}, {'id': 6, 'type': 'UltimateSDUpscale', 'pos': [800, 0], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'image', 'type': 'IMAGE', 'link': 3}, {'name': 'model', 'type': 'MODEL', 'link': 4}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 5}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 6}, {'name': 'vae', 'type': 'VAE', 'link': 7}, {'name': 'upscale_model', 'type': 'UPSCALE_MODEL', 'link': 8}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [9]}], 'properties': {'Node name for S&R': 'UltimateSDUpscale'}, 'widgets_values': [2.0, 42, 'fixed', 20, 5.0, 'dpmpp_2m', 'karras', 0.2, 'Linear', 768, 768, 8, 64, 'Half Tile', 0.15, 64, 8, 32, True, True, 1], 'title': 'Creative 2x — tiles can change faces and lettering'}, {'id': 7, 'type': 'SaveImage', 'pos': [1200, 0], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 9}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Creative']}], 'links': [[1, 1, 1, 2, 0, 'CLIP'], [2, 1, 1, 3, 0, 'CLIP'], [3, 4, 0, 6, 0, 'IMAGE'], [4, 1, 0, 6, 1, 'MODEL'], [5, 2, 0, 6, 2, 'CONDITIONING'], [6, 3, 0, 6, 3, 'CONDITIONING'], [7, 1, 2, 6, 4, 'VAE'], [8, 5, 0, 6, 5, 'UPSCALE_MODEL'], [9, 6, 0, 7, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Restore SeedVR2': {'last_node_id': 11, 'last_link_id': 15, 'nodes': [{'id': 1, 'type': 'LoadImage', 'pos': [0, 100], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [1]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'Upload an image to enlarge'}, {'id': 2, 'type': 'ImageScaleBy', 'pos': [360, 100], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [{'name': 'image', 'type': 'IMAGE', 'link': 1}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [2, 14]}], 'properties': {'Node name for S&R': 'ImageScaleBy'}, 'widgets_values': ['lanczos', 2.0], 'title': '2x target; start with source at 1024px or smaller'}, {'id': 3, 'type': 'SeedVR2Preprocess', 'pos': [720, 100], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'resized_images', 'type': 'IMAGE', 'link': 2}], 'outputs': [{'name': 'images', 'type': 'IMAGE', 'links': [3]}], 'properties': {'Node name for S&R': 'SeedVR2Preprocess'}, 'widgets_values': []}, {'id': 4, 'type': 'VAELoader', 'pos': [360, 440], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'VAE', 'type': 'VAE', 'links': [4, 12]}], 'properties': {'Node name for S&R': 'VAELoader'}, 'widgets_values': ['seedvr2_ema_vae_fp16.safetensors']}, {'id': 5, 'type': 'UNETLoader', 'pos': [0, 440], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'MODEL', 'type': 'MODEL', 'links': [5, 7]}], 'properties': {'Node name for S&R': 'UNETLoader'}, 'widgets_values': ['seedvr2_3b_int8_convrot.safetensors', 'default']}, {'id': 6, 'type': 'VAEEncodeTiled', 'pos': [1080, 100], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'pixels', 'type': 'IMAGE', 'link': 3}, {'name': 'vae', 'type': 'VAE', 'link': 4}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [6, 10]}], 'properties': {'Node name for S&R': 'VAEEncodeTiled'}, 'widgets_values': [512, 128, 64, 8]}, {'id': 7, 'type': 'SeedVR2Conditioning', 'pos': [1080, 440], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 5}, {'name': 'vae_conditioning', 'type': 'LATENT', 'link': 6}], 'outputs': [{'name': 'positive', 'type': 'CONDITIONING', 'links': [8]}, {'name': 'negative', 'type': 'CONDITIONING', 'links': [9]}], 'properties': {'Node name for S&R': 'SeedVR2Conditioning'}, 'widgets_values': []}, {'id': 8, 'type': 'KSampler', 'pos': [1440, 100], 'size': [320, 250], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'name': 'model', 'type': 'MODEL', 'link': 7}, {'name': 'positive', 'type': 'CONDITIONING', 'link': 8}, {'name': 'negative', 'type': 'CONDITIONING', 'link': 9}, {'name': 'latent_image', 'type': 'LATENT', 'link': 10}], 'outputs': [{'name': 'LATENT', 'type': 'LATENT', 'links': [11]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'fixed', 1, 1.0, 'euler', 'simple', 1.0]}, {'id': 9, 'type': 'VAEDecodeTiled', 'pos': [1800, 100], 'size': [320, 250], 'flags': {}, 'order': 8, 'mode': 0, 'inputs': [{'name': 'samples', 'type': 'LATENT', 'link': 11}, {'name': 'vae', 'type': 'VAE', 'link': 12}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [13]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 128, 64, 8]}, {'id': 10, 'type': 'SeedVR2PostProcessing', 'pos': [2160, 100], 'size': [320, 250], 'flags': {}, 'order': 9, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 13}, {'name': 'original_resized_images', 'type': 'IMAGE', 'link': 14}], 'outputs': [{'name': 'images', 'type': 'IMAGE', 'links': [15]}], 'properties': {'Node name for S&R': 'SeedVR2PostProcessing'}, 'widgets_values': ['none']}, {'id': 11, 'type': 'SaveImage', 'pos': [2520, 100], 'size': [320, 250], 'flags': {}, 'order': 10, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 15}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/Restore SeedVR2']}], 'links': [[1, 1, 0, 2, 0, 'IMAGE'], [2, 2, 0, 3, 0, 'IMAGE'], [3, 3, 0, 6, 0, 'IMAGE'], [4, 4, 0, 6, 1, 'VAE'], [5, 5, 0, 7, 0, 'MODEL'], [6, 6, 0, 7, 1, 'LATENT'], [7, 5, 0, 8, 0, 'MODEL'], [8, 7, 0, 8, 1, 'CONDITIONING'], [9, 7, 1, 8, 2, 'CONDITIONING'], [10, 6, 0, 8, 3, 'LATENT'], [11, 8, 0, 9, 0, 'LATENT'], [12, 4, 0, 9, 1, 'VAE'], [13, 9, 0, 10, 0, 'IMAGE'], [14, 2, 0, 10, 1, 'IMAGE'], [15, 10, 0, 11, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Face Swap': {'last_node_id': 8, 'last_link_id': 7, 'nodes': [{'id': 1, 'type': 'LoadImage', 'pos': [0, 100], 'size': [320, 250], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [1]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'SOURCE — the face identity to use'}, {'id': 2, 'type': 'LoadImage', 'pos': [0, 500], 'size': [320, 250], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [2]}, {'name': 'MASK', 'type': 'MASK', 'links': []}], 'properties': {'Node name for S&R': 'LoadImage'}, 'widgets_values': ['', 'image'], 'title': 'TARGET — photo to change'}, {'id': 3, 'type': 'StudioFaceDetect', 'pos': [380, 100], 'size': [320, 250], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [{'name': 'image', 'type': 'IMAGE', 'link': 1}], 'outputs': [{'name': 'faces', 'type': 'STUDIO_FACES', 'links': [5]}, {'name': 'detected_faces', 'type': 'STRING', 'links': [3]}], 'properties': {'Node name for S&R': 'StudioFaceDetect'}, 'widgets_values': [], 'title': 'Detect source faces — left to right'}, {'id': 4, 'type': 'StudioFaceDetect', 'pos': [380, 500], 'size': [320, 250], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [{'name': 'image', 'type': 'IMAGE', 'link': 2}], 'outputs': [{'name': 'faces', 'type': 'STUDIO_FACES', 'links': [6]}, {'name': 'detected_faces', 'type': 'STRING', 'links': [4]}], 'properties': {'Node name for S&R': 'StudioFaceDetect'}, 'widgets_values': [], 'title': 'Detect target faces — left to right'}, {'id': 5, 'type': 'PreviewAny', 'pos': [780, 50], 'size': [320, 250], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'name': 'source', 'type': '*', 'link': 3}], 'outputs': [{'name': 'STRING', 'type': 'STRING', 'links': []}], 'properties': {'Node name for S&R': 'PreviewAny'}, 'widgets_values': [], 'title': 'Detected source faces'}, {'id': 6, 'type': 'PreviewAny', 'pos': [780, 500], 'size': [320, 250], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'name': 'source', 'type': '*', 'link': 4}], 'outputs': [{'name': 'STRING', 'type': 'STRING', 'links': []}], 'properties': {'Node name for S&R': 'PreviewAny'}, 'widgets_values': [], 'title': 'Detected target faces'}, {'id': 7, 'type': 'StudioFaceSwap', 'pos': [1180, 200], 'size': [320, 250], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [{'name': 'source_faces', 'type': 'STUDIO_FACES', 'link': 5}, {'name': 'target_faces', 'type': 'STUDIO_FACES', 'link': 6}], 'outputs': [{'name': 'IMAGE', 'type': 'IMAGE', 'links': [7]}], 'properties': {'Node name for S&R': 'StudioFaceSwap'}, 'widgets_values': [0, 0], 'title': 'Select source and target face indices (0 = leftmost)'}, {'id': 8, 'type': 'SaveImage', 'pos': [1580, 200], 'size': [320, 250], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 7}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/FaceSwap']}], 'links': [[1, 1, 0, 3, 0, 'IMAGE'], [2, 2, 0, 4, 0, 'IMAGE'], [3, 3, 1, 5, 0, 'STRING'], [4, 4, 1, 6, 0, 'STRING'], [5, 3, 0, 7, 0, 'STUDIO_FACES'], [6, 4, 0, 7, 1, 'STUDIO_FACES'], [7, 7, 0, 8, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.65, 'offset': [180, 180]}}, 'version': 0.4}, 'Create FLUX3': {'last_node_id': 30, 'last_link_id': 91, 'nodes': [{'id': 17, 'type': 'PrimitiveStringMultiline', 'pos': [-680, -950], 'size': [490, 450], 'flags': {}, 'order': 13, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'STRING', 'name': 'STRING', 'type': 'STRING', 'links': [17]}], 'properties': {'Node name for S&R': 'PrimitiveStringMultiline'}, 'widgets_values': ['A cinematic photograph of a lighthouse above a stormy coastline at sunset.']}, {'id': 12, 'type': 'CLIPLoader', 'pos': [-140, -930], 'size': [500, 110], 'flags': {}, 'order': 10, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'CLIP', 'name': 'CLIP', 'type': 'CLIP', 'links': [10]}], 'properties': {'Node name for S&R': 'CLIPLoader', 'models': [{'name': 'qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors', 'url': 'https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/main/text_encoders/qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors', 'directory': 'text_encoders'}]}, 'widgets_values': ['qwen3.5_9b_flux3_prompt_upsampler_w4a8.safetensors', 'flux3', 'default']}, {'id': 14, 'type': 'PrimitiveStringMultiline', 'pos': [-150, -750], 'size': [530, 700], 'flags': {}, 'order': 0, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'STRING', 'name': 'STRING', 'type': 'STRING', 'links': [13]}], 'title': 'Text (System prompt)', 'properties': {'Node name for S&R': 'PrimitiveStringMultiline'}, 'widgets_values': ['You are the prompt upsampler of a text-to-image model. Expand the user\'s short request into ONE detailed, self-contained caption: the exact text the image model receives. Output only the caption, with no greetings, explanations, refusals, questions, Markdown, headings, JSON, coordinates, ids or angle-bracket tokens, and no quotes around it. If the input is not clearly an image request, caption a visualization of its concept. The request may begin with a "TARGET CANVAS: W:H" line: compose for that canvas (wide spreads content sideways, tall stacks it) and never mention the ratio or the canvas.\n\nFidelity and planning: keep every explicit subject, attribute, count, spatial constraint, style cue and piece of wording; never contradict, substitute, drop or narrow the request, and keep an abstract concept abstract. Decide everything left open as ONE concrete, consistent plan; never offer alternatives or hedge. Never re-express a stated constraint through another mechanism; add camera optics (depth of field, bokeh, blur) only when asked for or implied. Without a stated style, plan an actual photograph and open with the medium convention ("Full-color outdoor photograph, ...", "Full-color indoor photograph, ...", "Studio photograph, ...", "Close-up photograph, ..."); keep "Photorealistic" for renders, and turn "photorealistic" or "realistic" requests for photos into the photograph convention. Give a photograph a few ordinary, non-ideal specifics drawn from the scene\'s own objects instead of the stock-photo ideal. Do not invent names, brands, logos or celebrities; keep user-named ones. A long request keeps its content but not its wording: drop appraisal and sales vocabulary, turn moods into observable physical facts, and add only new concrete content with positions. Time of day is lighting, not rendered text.\n\nCaption format: one continuous prose block in the generative voice, never "this image shows", "we can see" or "is visible". Lead with medium and style, then scene, lighting and camera, then the subjects with concrete colors, materials and state. Every person in focus gets an observable profile: ethnicity, gender, age range, skin tone ("fair skin", "brown skin"), hair and clothing. Make the layout reconstructable from words: frame-relative positions (frame-left, frame-right, upper, lower, foreground, midground, background), sentence-initial anchors ("On the left, ...", "In the background, ..."), relations (holds, faces, behind) and a consistent order; describe up to about ten instances one by one, beyond that give the exact count and arrangement. Out-of-focus or distant planes are shapes, colors and silhouettes, never identities or readable text, and the background takes about a sixth of the caption. Be definite: no hedging adverbs, no quality or emotion judgments, no closing sentence about mood; state negations as what is present. Rendered text: transcribe every wording verbatim inside double quotes in its original language and script, keep the line breaks of multi-line text, and give its typography, color and placement; author concrete, domain-correct wording when the request needs text it does not supply, and name canonical structures (calendar, keyboard, periodic table) exactly with their full layout. All prose is English; only quoted text keeps its language. Typical length is 100 to 350 words, more for complex scenes.\n\nOutput only the caption.\n']}, {'id': 11, 'type': 'TextGenerate', 'pos': [400, -930], 'size': [400, 460], 'flags': {}, 'order': 9, 'mode': 0, 'inputs': [{'localized_name': 'clip', 'name': 'clip', 'type': 'CLIP', 'link': 10}, {'localized_name': 'image', 'name': 'image', 'shape': 7, 'type': 'IMAGE', 'link': None}, {'localized_name': 'video', 'name': 'video', 'shape': 7, 'type': 'IMAGE', 'link': None}, {'localized_name': 'audio', 'name': 'audio', 'shape': 7, 'type': 'AUDIO', 'link': None}, {'localized_name': 'system_prompt', 'name': 'system_prompt', 'shape': 7, 'type': 'STRING', 'link': 13}, {'localized_name': 'prompt', 'name': 'prompt', 'type': 'STRING', 'widget': {'name': 'prompt'}, 'link': 17}], 'outputs': [{'localized_name': 'generated_text', 'name': 'generated_text', 'type': 'STRING', 'links': [90]}, {'localized_name': 'thinking', 'name': 'thinking', 'type': 'STRING', 'links': []}], 'properties': {'Node name for S&R': 'TextGenerate'}, 'widgets_values': ['', 3584, 'on', 1, 1, 0.95, 0, 1.05, 0, 1.5, False, True, 'off']}, {'id': 1, 'type': 'UNETLoader', 'pos': [-150, 60], 'size': [340, 90], 'flags': {}, 'order': 1, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'MODEL', 'name': 'MODEL', 'type': 'MODEL', 'slot_index': 0, 'links': [1]}], 'properties': {'Node name for S&R': 'UNETLoader', 'models': [{'name': 'flux3-image-klein-int8-convrot.safetensors', 'url': 'https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/main/diffusion_models/flux3-image-klein-int8-convrot.safetensors', 'directory': 'diffusion_models'}]}, 'widgets_values': ['flux3-image-klein-int8-convrot.safetensors', 'default'], 'color': '#223', 'bgcolor': '#335'}, {'id': 4, 'type': 'CLIPTextEncode', 'pos': [250, 90], 'size': [450, 460], 'flags': {}, 'order': 4, 'mode': 0, 'inputs': [{'localized_name': 'clip', 'name': 'clip', 'type': 'CLIP', 'link': 2}, {'localized_name': 'text', 'name': 'text', 'type': 'STRING', 'widget': {'name': 'text'}, 'link': 20}], 'outputs': [{'localized_name': 'CONDITIONING', 'name': 'CONDITIONING', 'type': 'CONDITIONING', 'slot_index': 0, 'links': [8]}], 'title': 'Positive Prompt', 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': ['A close-up photograph of a red fox standing in fresh snow at sunrise. The fox faces the camera with alert ears and amber eyes, its thick orange fur dusted with snowflakes. Soft golden light comes from the left, the background is a blurred line of snow-covered pine trees, shallow depth of field.'], 'color': '#232', 'bgcolor': '#353'}, {'id': 2, 'type': 'CLIPLoader', 'pos': [-150, 220], 'size': [340, 110], 'flags': {}, 'order': 2, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'CLIP', 'name': 'CLIP', 'type': 'CLIP', 'slot_index': 0, 'links': [2, 3]}], 'properties': {'Node name for S&R': 'CLIPLoader', 'models': [{'name': 'qwen3vl_4b_int8_convrot.safetensors', 'url': 'https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/main/text_encoders/qwen3vl_4b_int8_convrot.safetensors', 'directory': 'text_encoders'}]}, 'widgets_values': ['qwen3vl_4b_int8_convrot.safetensors', 'flux3', 'default'], 'color': '#432', 'bgcolor': '#653'}, {'id': 7, 'type': 'KSampler', 'pos': [760, 30], 'size': [320, 480], 'flags': {}, 'order': 7, 'mode': 0, 'inputs': [{'localized_name': 'model', 'name': 'model', 'type': 'MODEL', 'link': 1}, {'localized_name': 'positive', 'name': 'positive', 'type': 'CONDITIONING', 'link': 8}, {'localized_name': 'negative', 'name': 'negative', 'type': 'CONDITIONING', 'link': 9}, {'localized_name': 'latent_image', 'name': 'latent_image', 'type': 'LATENT', 'link': 39}], 'outputs': [{'localized_name': 'LATENT', 'name': 'LATENT', 'type': 'LATENT', 'slot_index': 0, 'links': [4]}], 'properties': {'Node name for S&R': 'KSampler'}, 'widgets_values': [42, 'randomize', 16, 1.0, 'euler', 'simple', 1.0]}, {'id': 3, 'type': 'VAELoader', 'pos': [-150, 400], 'size': [340, 60], 'flags': {}, 'order': 3, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'VAE', 'name': 'VAE', 'type': 'VAE', 'slot_index': 0, 'links': [5]}], 'properties': {'Node name for S&R': 'VAELoader', 'models': [{'name': 'flux2-vae.safetensors', 'url': 'https://huggingface.co/black-forest-labs/flux-3-image-klein-comfy/resolve/main/vae/flux2-vae.safetensors', 'directory': 'vae'}]}, 'widgets_values': ['flux2-vae.safetensors'], 'color': '#323', 'bgcolor': '#535'}, {'id': 13, 'type': 'PreviewAny', 'pos': [830, -660], 'size': [350, 490], 'flags': {}, 'order': 11, 'mode': 0, 'inputs': [{'localized_name': 'source', 'name': 'source', 'type': '*', 'link': 90}], 'outputs': [{'localized_name': 'STRING', 'name': 'STRING', 'type': 'STRING', 'links': [20]}], 'properties': {'Node name for S&R': 'PreviewAny'}, 'widgets_values': []}, {'id': 6, 'type': 'EmptyFlux2LatentImage', 'pos': [-120, 590], 'size': [300, 110], 'flags': {}, 'order': 6, 'mode': 0, 'inputs': [], 'outputs': [{'localized_name': 'LATENT', 'name': 'LATENT', 'type': 'LATENT', 'slot_index': 0, 'links': [39]}], 'properties': {'Node name for S&R': 'EmptyFlux2LatentImage'}, 'widgets_values': [1024, 1024, 1], 'color': '#323', 'bgcolor': '#535'}, {'id': 5, 'type': 'CLIPTextEncode', 'pos': [250, 610], 'size': [460, 110], 'flags': {}, 'order': 5, 'mode': 0, 'inputs': [{'localized_name': 'clip', 'name': 'clip', 'type': 'CLIP', 'link': 3}], 'outputs': [{'localized_name': 'CONDITIONING', 'name': 'CONDITIONING', 'type': 'CONDITIONING', 'slot_index': 0, 'links': [9]}], 'title': 'Negative Prompt (empty = reference default)', 'properties': {'Node name for S&R': 'CLIPTextEncode'}, 'widgets_values': [''], 'color': '#223', 'bgcolor': '#335'}, {'id': 8, 'type': 'VAEDecodeTiled', 'pos': [1120, 30], 'size': [210, 50], 'flags': {}, 'order': 8, 'mode': 0, 'inputs': [{'localized_name': 'samples', 'name': 'samples', 'type': 'LATENT', 'link': 4}, {'localized_name': 'vae', 'name': 'vae', 'type': 'VAE', 'link': 5}], 'outputs': [{'localized_name': 'IMAGE', 'name': 'IMAGE', 'type': 'IMAGE', 'slot_index': 0, 'links': [91]}], 'properties': {'Node name for S&R': 'VAEDecodeTiled'}, 'widgets_values': [512, 64, 64, 8]}, {'id': 30, 'type': 'SaveImage', 'pos': [1400, 30], 'size': [320, 250], 'flags': {}, 'order': 30, 'mode': 0, 'inputs': [{'name': 'images', 'type': 'IMAGE', 'link': 91}], 'outputs': [], 'properties': {'Node name for S&R': 'SaveImage'}, 'widgets_values': ['ImageStudio/FLUX3']}], 'links': [[2, 2, 0, 4, 0, 'CLIP'], [20, 13, 0, 4, 1, 'STRING'], [1, 1, 0, 7, 0, 'MODEL'], [8, 4, 0, 7, 1, 'CONDITIONING'], [9, 5, 0, 7, 2, 'CONDITIONING'], [3, 2, 0, 5, 0, 'CLIP'], [4, 7, 0, 8, 0, 'LATENT'], [5, 3, 0, 8, 1, 'VAE'], [39, 6, 0, 7, 3, 'LATENT'], [10, 12, 0, 11, 0, 'CLIP'], [13, 14, 0, 11, 4, 'STRING'], [17, 17, 0, 11, 5, 'STRING'], [90, 11, 0, 13, 0, 'STRING'], [91, 8, 0, 30, 0, 'IMAGE']], 'groups': [], 'config': {}, 'extra': {'ds': {'scale': 0.5, 'offset': [750, 1000]}}, 'version': 0.4}}
COMFY_REV='1332fabfcc507aa7c548cd5152071ed494c207e8'
MODEL='Juggernaut-XL_v9_RunDiffusionPhoto_v2.safetensors'
MODEL_SHA='c9e3e68f89b8e38689e1097d4be4573cf308de4e3fd044c64ca697bdb4aa8bca'
MODEL_SIZE=7105348188
CF_SHA='1f99fb518b83dec2ba8d870a7cc991d82a77e6196295a3d2754ba17ab9f91872'
# Renderer dependencies are unchanged; reuse a retained v4 installation.
BASE=Path('/kaggle/temp/image-studio-v4'); BASE.mkdir(parents=True,exist_ok=True)
OUT=Path('/kaggle/working/image-studio'); OUT.mkdir(parents=True,exist_ok=True)
lock=open(BASE/'startup.lock','w')
try: fcntl.flock(lock,fcntl.LOCK_EX|fcntl.LOCK_NB)
except BlockingIOError: raise RuntimeError('Image Studio is already running. Stop its startup cell before starting again.')
children=[]
# Fail before downloading if the previous studio still owns this port.
import socket
probe=socket.socket()
try: probe.bind(('127.0.0.1',8188))
except OSError: raise RuntimeError('Port 8188 is occupied. Stop the earlier studio cell before starting v5.')
finally: probe.close()

def run(args,**kw):subprocess.run(list(map(str,args)),check=True,**kw)
def sha(path):
 h=hashlib.sha256()
 with open(path,'rb') as f:
  for chunk in iter(lambda:f.read(8*1024*1024),b''):h.update(chunk)
 return h.hexdigest()
def download(url,path,expected,size=None):
 if path.exists() and sha(path)==expected:return
 if path.exists():raise RuntimeError(f'Checksum mismatch: {path}. Preserve or move this file before retrying.')
 if size and shutil.disk_usage(path.parent).free<size+2*1024**3:raise RuntimeError('Insufficient free disk for model plus 2 GiB reserve.')
 temp=path.with_suffix(path.suffix+'.partial')
 print('Downloading',path.name,flush=True)
 with urllib.request.urlopen(url,timeout=120) as response,open(temp,'wb') as target:shutil.copyfileobj(response,target,8*1024*1024)
 if sha(temp)!=expected:raise RuntimeError(f'Download checksum mismatch: {temp}')
 temp.replace(path)
def spawn(args,log):
 f=open(log,'w');p=subprocess.Popen(list(map(str,args)),stdout=f,stderr=subprocess.STDOUT,start_new_session=True);f.close();children.append(p);return p
# Prepare only missing face models; keep generation packs attached and untouched.
try:
 resolved_models=prepare_startup_models(
  model_assets(LOAD_SDXL,ENABLE_SEEDVR2,ENABLE_FLUX3,ENABLE_FACESWAP),PACK_BUILDER_SOURCE,
  require_attached=REQUIRE_ATTACHED_MODELS,prepare_faces=AUTO_PREPARE_FACE_PACK,
  save_face_dataset=AUTO_SAVE_PREPARED_FACE_PACK)
except BaseException:
 fcntl.flock(lock,fcntl.LOCK_UN);lock.close();raise
try:
 import torch
 if not torch.cuda.is_available():raise RuntimeError('Select a Kaggle GPU accelerator and restart this cell.')
 print('GPU:',torch.cuda.get_device_name(0),'Torch:',torch.__version__,flush=True)
 free,total=torch.cuda.mem_get_info(0)
 if free<10*1024**3:raise RuntimeError('GPU 0 has less than 10 GiB free. Stop other GPU notebooks/processes first.')
 # Separate T4s do not form one shared VRAM pool. This studio uses GPU 0.
 comfy=BASE/'ComfyUI'
 if not comfy.exists():
  run(['git','clone','--no-checkout','https://github.com/comfyanonymous/ComfyUI.git',comfy])
  run(['git','-C',comfy,'checkout','--detach',COMFY_REV])
 if subprocess.check_output(['git','-C',str(comfy),'rev-parse','HEAD'],text=True).strip()!=COMFY_REV:raise RuntimeError('Unexpected ComfyUI revision.')
 venv=BASE/'venv';python=venv/'bin/python'
 if not python.exists():run([sys.executable,'-m','venv','--system-site-packages','--without-pip',venv])
 constraints=BASE/'constraints.txt'
 import importlib.metadata as meta
 protected=['torch','torchvision','torchaudio','numpy','scipy']
 constraints.write_text('\n'.join(f'{p}=={meta.version(p)}' for p in protected))
 requirements=BASE/'requirements.txt'
 requirements.write_text('\n'.join(x for x in (comfy/'requirements.txt').read_text().splitlines() if x.strip() not in protected)+'\ntransformers==5.22.2\nhuggingface-hub==2.4.0\n')
 fingerprint=hashlib.sha256((requirements.read_text()+constraints.read_text()+sys.version).encode()).hexdigest()
 marker=BASE/'setup.complete'
 if not marker.exists() or marker.read_text()!=fingerprint:
  run([sys.executable,'-m','pip','--python',python,'install','-r',requirements,'-c',constraints])
  run([python,'-c','import torch, torchvision, transformers, aiohttp, safetensors; assert torch.cuda.is_available()'])
  marker.write_text(fingerprint)
 if ENABLE_FACESWAP:
  face_marker=BASE/'face-setup.complete'
  if not face_marker.exists() or face_marker.read_text()!=fingerprint:
   env=os.environ.copy()
   if shutil.which('gcc'):env['CC']='gcc'
   if shutil.which('g++'):env['CXX']='g++'
   run([sys.executable,'-m','pip','--python',python,'install','insightface==0.7.3','onnxruntime==1.23.2','-c',constraints],env=env)
   run([python,'-c','import insightface,onnxruntime,cv2'])
   face_marker.write_text(fingerprint)
  custom=comfy/'custom_nodes/image_studio_faces';custom.mkdir(parents=True,exist_ok=True)
  (custom/'__init__.py').write_text(FACE_NODE_SOURCE)
 run([sys.executable,'-m','pip','--python',python,'freeze'],stdout=open(OUT/'packages.txt','w'))
 ultimate=comfy/'custom_nodes/ComfyUI_UltimateSDUpscale'
 if LOAD_SDXL:
  if not ultimate.exists():
   run(['git','clone','--no-checkout','https://github.com/ssitu/ComfyUI_UltimateSDUpscale.git',ultimate])
   run(['git','-C',ultimate,'checkout','--detach','a5547db9e1d07d3318bb21e9e9c474f4c1e9c8df'])
  if subprocess.check_output(['git','-C',str(ultimate),'rev-parse','HEAD'],text=True).strip()!='a5547db9e1d07d3318bb21e9e9c474f4c1e9c8df':raise RuntimeError('Unexpected Ultimate SD Upscale revision.')
  run(['git','-C',ultimate,'submodule','update','--init','--recursive'])
 assets=model_assets(LOAD_SDXL,ENABLE_SEEDVR2,ENABLE_FLUX3,ENABLE_FACESWAP)
 for folder,filename,url,digest,size in assets:
  dest=comfy/'models'/folder/filename;dest.parent.mkdir(parents=True,exist_ok=True)
  verified=False
  if not dest.exists():
   candidates=resolved_models.get(filename,[])
   for candidate in candidates:
    if candidate.is_file() and candidate.stat().st_size==size and sha(candidate)==digest:
     dest.symlink_to(candidate);verified=True;print('Reusing verified model:',candidate,flush=True);break
  if REQUIRE_ATTACHED_MODELS and not dest.exists():
   raise RuntimeError('Missing saved model: '+filename+'. Attach your saved Image Studio Model Packs output via Add Input. No model was downloaded.')
  if ENABLE_FLUX3 and 'black-forest-labs' in url and not dest.exists():
   if shutil.disk_usage(dest.parent).free<size+2*1024**3:raise RuntimeError('Insufficient disk space for selected FLUX3 assets.')
   from kaggle_secrets import UserSecretsClient
   try: token=UserSecretsClient().get_secret('HF_TOKEN')
   except Exception: raise RuntimeError('Enable Kaggle Secret HF_TOKEN with access to the accepted FLUX 3 model terms.') from None
   env=os.environ.copy();env['HF_TOKEN']=token
   try:
    code="import os;from huggingface_hub import hf_hub_download;hf_hub_download(repo_id='black-forest-labs/flux-3-image-klein-comfy',filename="+repr(folder+'/'+filename)+",revision='9d36114cf7a2e03a45c28f5cc4987d20ebfdb6c1',local_dir="+repr(str(comfy/'models'))+",token=os.environ['HF_TOKEN'])"
    run([python,'-c',code],env=env)
   finally:env.pop('HF_TOKEN',None);del token
  if not verified:download(url,dest,digest,size)
 if ENABLE_FACESWAP:
  import zipfile,zlib
  archive=comfy/'models/insightface/buffalo_l.zip'
  folder=comfy/'models/insightface/models/buffalo_l';folder.mkdir(parents=True,exist_ok=True)
  with zipfile.ZipFile(archive) as z:
   allowed={'1k3d68.onnx','2d106det.onnx','det_10g.onnx','genderage.onnx','w600k_r50.onnx'}
   for item in z.infolist():
    if item.is_dir():continue
    if item.filename not in allowed:raise RuntimeError('Unexpected face model archive member: '+item.filename)
    target=folder/item.filename
    valid=target.exists() and target.stat().st_size==item.file_size
    if valid:
     crc=0
     with open(target,'rb') as f:
      for block in iter(lambda:f.read(8*1024*1024),b''):crc=zlib.crc32(block,crc)
     valid=(crc & 0xffffffff)==item.CRC
    if not valid:
     tmp=target.with_suffix('.partial')
     with z.open(item) as src,open(tmp,'wb') as dst:shutil.copyfileobj(src,dst)
     tmp.replace(target)
   if not all((folder/name).is_file() for name in allowed):raise RuntimeError('Incomplete buffalo_l model archive.')
 # Only install workflows for enabled model families.
 WORKFLOW_PAYLOAD={k:v for k,v in WORKFLOW_PAYLOAD.items() if (LOAD_SDXL or k in ['Upscale 2x','Upscale 4x','Restore SeedVR2','Create FLUX3','Face Swap']) and (ENABLE_FACESWAP or k!='Face Swap') and (ENABLE_FLUX3 or k!='Create FLUX3') and (ENABLE_SEEDVR2 or k!='Restore SeedVR2')}
 workflow_dir=comfy/'user/default/workflows';workflow_dir.mkdir(parents=True,exist_ok=True)
 for name,wf in WORKFLOW_PAYLOAD.items():(workflow_dir/f'Image Studio - {name}.json').write_text(json.dumps(wf,indent=2))
 # Stable output destination: PNGs retain the actual graph, prompt, seed and settings.
 (OUT/'provenance.json').write_text(json.dumps({'comfy_revision':COMFY_REV,'model':MODEL,'sha256':MODEL_SHA,'assets':[{'file':f,'sha256':d,'bytes':z} for _,f,_,d,z in assets],'torch':torch.__version__},indent=2))
 cf=BASE/'cloudflared'
 download('https://github.com/cloudflare/cloudflared/releases/download/2026.10.7/cloudflared-linux-amd64',cf,CF_SHA)
 cf.chmod(0o700)
 server=spawn([python,comfy/'main.py','--listen','127.0.0.1','--port','8188','--lowvram','--force-fp16','--preview-method','none','--output-directory',OUT/'images','--disable-auto-launch'],OUT/'comfy.log')
 deadline=time.monotonic()+180
 while True:
  if server.poll() is not None:raise RuntimeError('ComfyUI stopped. See comfy.log in Kaggle outputs.')
  try:
   with urllib.request.urlopen('http://127.0.0.1:8188/object_info',timeout=3) as r:info=json.load(r)
   break
  except OSError:
   if time.monotonic()>deadline:raise RuntimeError('ComfyUI startup timed out. See comfy.log.')
   time.sleep(1)
 # Validate installed node classes before offering the browser link.
 for wf in WORKFLOW_PAYLOAD.values():
  for node in wf['nodes']:
   if node['type'] not in info:raise RuntimeError('Required node unavailable: '+node['type'])
 tunnel=spawn([cf,'tunnel','--url','http://127.0.0.1:8188','--no-autoupdate'],OUT/'tunnel.log')
 deadline=time.monotonic()+90
 url=None
 while time.monotonic()<deadline:
  text=(OUT/'tunnel.log').read_text()
  matches=re.findall(r'https://[a-z0-9-]+\.trycloudflare\.com',text)
  if matches:url=matches[-1];break
  if tunnel.poll() is not None:raise RuntimeError('Cloudflare stopped; inspect tunnel.log.')
  time.sleep(1)
 if not url:raise RuntimeError('No Cloudflare URL received; inspect tunnel.log.')
 print('\nOPEN IMAGE STUDIO:',url,flush=True)
 print('Workflow menu → Image Studio: '+', '.join(WORKFLOW_PAYLOAD)+'. Select a workflow, then press Run.')
 print('No generation has started. Keep this cell running. Anyone with this session URL can access the studio.')
 print('Save Version with outputs before ending Kaggle. Images:',OUT/'images',flush=True)
 while server.poll() is None and tunnel.poll() is None:time.sleep(2)
 raise RuntimeError('Studio or tunnel stopped; see the saved logs.')
finally:
 for p in reversed(children):
  if p.poll() is None:
   os.killpg(p.pid,signal.SIGTERM)
   try:p.wait(timeout=10)
   except subprocess.TimeoutExpired:os.killpg(p.pid,signal.SIGKILL);p.wait()
 fcntl.flock(lock,fcntl.LOCK_UN);lock.close()
```
